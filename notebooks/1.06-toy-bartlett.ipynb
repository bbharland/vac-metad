# Random draw of the toy ACF

```python
import numpy as np
import matplotlib.pyplot as plt
from vac_metad.acf.toy import toy_model, bartlett_draws


def plot_acf_bartlett(acf, bartlett):
    fig, ax = plt.subplots(figsize=(8, 5))
    ax.plot(acf.t, acf.ct, "k:", lw=1.8, label="exact $C(\\tau)$")

    for d in bartlett.draws:
        ax.plot(acf.t, d, color="tab:blue", lw=0.4, alpha=0.5)
    ax.plot(acf.t, bartlett.mean, color="tab:blue", lw=1.8, label="mean (with bias)")
    ax.plot([], [], color="tab:blue", lw=0.4, label=f"{len(bartlett.draws)} draws")

    ax.axhline(0, color="gray", lw=0.5)
    ax.set_xlabel(r"$\tau$ (ps)")
    ax.set_ylabel(r"$\hat C(\tau)$")
    ax.legend(frameon=False)
    fig.tight_layout()
    plt.show()


acf = toy_model(frametime=1.0, num_lags=20_000, num_frames=3_000_000)
bartlett = bartlett_draws(acf, num_draws=20)

plot_acf_bartlett(acf, bartlett)
```
